### 2.Pivot Table

```python
pd.pivot_table(data = insurance_data,values="charges",index="region",columns="sex") #pivot_table comparing to categarical features vs 1 continuous data
```

```python
pd.pivot_table(data = insurance_data,values="charges",index="region",columns=["sex","children"]) #pivot_table comparing to categarical features vs 1 continuous data
```

#### 6.1 Create a new column in the same dataframe with the name "Eligibility_Criteriya" where if age is >20, say Eligible else not

```python
def get_age(x):
    if x > 20:
        return "Eligible"
    else:
        return "Not Eligible"

insurance_data["Eligibility_Criteriya"]= insurance_data["age"].apply(get_age)
insurance_data
```

### 7.lambda Function()

* Anonymous function/nameless function

```python
insurance_data["Eligibility_Criteriya"]= insurance_data["age"].apply(lambda x:"Eligible" if x>20 else "Not Eligible")
```

```python
insurance_data
```

```python
insurance_data["EC"]= insurance_data["age"].apply(lambda x:1  if x>20 else 0)
```
