## What is Abstraction?
## Definition (Interview)

* Abstraction is the process of hiding implementation details and showing only the essential features of an object to the user.

## Simple Meaning

* Think about driving a car.

* You use the steering wheel, accelerator, and brake.
* You don't need to know how the engine, gearbox, or fuel injection system works internally.

**This is abstraction:** you interact with what is necessary while the complex implementation is hidden.

## Real-Life Examples
* ATM Machine: You withdraw money without knowing how the bank processes the transaction.
* Mobile Phone: You make calls without understanding the internal communication network.
* TV Remote: You press a button to change channels without knowing the electronics inside.

```python
from abc import ABC, abstractmethod
```

```python
# Abstract Class
class Animal(ABC):

    @abstractmethod #this is a first abstact is always present in child class otherwise not proceed
    def sound(self):
        pass
        
#Child class
class Dog(Animal):

    def sound(self):
        print("Dog")

    def color(self):
        print("Brown")

#Create Object
dog = Dog()
dog.color() 
```

# Abstaction:

## The TV turns ON.

## Do you know what happens inside the TV?

## No.

### Maybe the TV:

* Checks electricity
* Starts the processor
* Loads the operating system
* Connects to the display
* Starts the speakers

### You don't see any of this.

### You only press one button.

### This is the main idea of **abstraction**.

```python
class TV:

    def check_power(self):
        print("Checking electricity")

    def start_processor(self):
        print("Processor Started")

    def start_screen(self):
        print("Screen Started")

    def power_on(self):
        self.check_power()
        self.start_processor()
        self.start_screen()

tv = TV()

tv.power_on()
```

# 2. Inheritance (Reuse Existing Code)

```python
class User:

    def __init__(self, name, pin):
        self.name = name
        self.pin = pin

    def login(self):
        print(f"{self.name} Logged In")

class Admin(User):

    def block_account(self):
        print("Account Blocked")


admin = Admin("Tamil", 1234)

admin.login()
admin.block_account()
```

# 3. Polymorphism (One Method, Different Behavior)

```python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


class Cow:

    def sound(self):
        print("Moo")


dog = Dog()
cat = Cat()
cow = Cow()

dog.sound()
cat.sound()
cow.sound()
```

# CONCEPT OF CLASS'S

| Concept          | Meaning                         | ATM Example                                                                                        |
| ---------------- | ------------------------------- | -------------------------------------------------------------------------------------------------- |
| **Abstraction**  | Hide complexity                 | User only clicks **Withdraw**, internal steps are hidden.                                          |
| **Inheritance**  | Reuse code                      | `Customer` inherits `login()` from `User`.                                                         |
| **Polymorphism** | Same method, different behavior | `withdraw()` behaves differently for `Customer` and `Admin`, or `interest()` differs across banks. |
