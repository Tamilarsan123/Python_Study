### 2.	Create a list of numbers and print only even numbers using a for loop.

```python
even_number = []

for i in range(1,100):
    if i%2 ==0:
        even_number.append(i)
print(even_number)
```

### 4.	Given a string, count how many vowels it has.

```python
user_input = input("Enter a vales:")
count = 0

for i in user_input.lower():
    
    if i in "aeiou":
        count+=1

print(count)
 
```

### 5.	Reverse a list without using built-in functions.

```python
my_list = [1,2,3,4,5,6,7,8,9]
new_list =[]

index_value = len(my_list)-1


while index_value>=0:
    new_list.append(my_list[index_value])
    index_value-=1
    
print(new_list)
    
    
    
```
