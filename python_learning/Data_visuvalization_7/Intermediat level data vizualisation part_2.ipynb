# Advanced level of Data Visualization

```python
import plotly.express as px

# Load a built-in dataset and filter for a specific year
df = px.data.gapminder().query("year == 2007 and continent == 'Europe'")

# Create interactive bar chart
fig = px.bar(
    df, 
    x="country", 
    y="pop", 
    title="Population of European Countries in 2007"
)

# Sort bars in descending order
fig.update_layout(xaxis={'categoryorder':'total descending'})

fig.show()
```

## NOTE:

1. Univariate charts can include more than one feature
2. Bivariate charts can include more than 2 feature


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Example: Generate some sample data
np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=50, scale=10, size=200)  # mean=50, std=10

# Convert to DataFrame for better handling
df = pd.DataFrame({'values': data})

# Validate that the column exists and contains numeric data
if 'values' not in df.columns or not pd.api.types.is_numeric_dtype(df['values']):
    raise ValueError("Data must contain a numeric column named 'values'.")

# Create histogram using seaborn.histplot
sns.set(style="whitegrid")  # Optional styling
sns.histplot(
    data=df,
    x='values',          # Column to plot
    bins=20,             # Number of bins
    kde=True,            # Overlay Kernel Density Estimate
    color='red',     # Bar color
    edgecolor='black'    # Border color for bars
)

# Add labels and title
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Values with Seaborn histplot")

# Show the plot
plt.show()
```
