## 1.Abstaction

## What is Abstraction?
## Definition (Interview)

* Abstraction is the process of hiding implementation details and showing only the essential features of an object to the user.

## Simple Meaning

* Think about driving a car.

* You use the steering wheel, accelerator, and brake.
* You don't need to know how the engine, gearbox, or fuel injection system works internally.

**This is abstraction:** you interact with what is necessary while the complex implementation is hidden.

## Real-Life Examples
* ATM Machine: You withdraw money without knowing how the bank processes the transaction.
* Mobile Phone: You make calls without understanding the internal communication network.
* TV Remote: You press a button to change channels without knowing the electronics inside.

### Example 1:

```python
from abc import ABC, abstractmethod
```

```python
# Abstract Class
class Animal(ABC):

    @abstractmethod #this is a first abstact is always present in child class otherwise not proceed
    def sound(self):
        pass
        
#Child class
class Dog(Animal):

    def sound(self):
        print("Dog")

    def color(self):
        print("Brown")

#Create Object
dog = Dog()
dog.sound() 
```

### Example-2

```python
from abc import ABC, abstractmethod

class TV(ABC):

    @abstractmethod
    def check_power(self):
        pass

    @abstractmethod
    def start_processor(self):
        pass

    @abstractmethod
    def start_screen(self):
        pass


class AirtelTV(TV):

    def check_power(self):
        print("Checking electricity")

    def start_processor(self):
        print("Processor Started")

    def start_screen(self):
        print("Screen Started")

    def power_on(self):
        self.check_power()
        self.start_processor()
        self.start_screen()


tv = AirtelTV()
tv.power_on()
```

## 2.Inheritance (reuse existing code)

* Inheritance is one of the four main concepts of Object-Oriented Programming (OOP).

### Simple Definition

* Inheritance means one class can use the properties and methods of another class.

* Think of it like a child inheriting qualities from its parents.

**Parent Class** (Base/Super Class) → Gives properties and methods.

**Child Class** (Derived/Sub Class) → Receives those properties and methods and can also add its own.

## The main types are:

**Single Inheritance**
  One child class inherits from one parent class.

**Multiple Inheritance**
 One child class inherits from more than one parent class.

**Multilevel Inheritance** 
 A class inherits from another derived class.

**Hierarchical Inheritance**
 Multiple child classes inherit from the same parent class.

**Hybrid Inheritance**
 A combination of two or more inheritance types (possible in Python because it supports multiple inheritance).

```python
class User:

    def __init__(self, name, pin):
        self.name = name
        self.pin = pin

    def login(self):
        print(f"{self.name} Logged In")

class Admin(User):

    def block_account(self):
        print("Account Blocked")


admin = Admin("Tamil", 1234)

admin.login()
admin.block_account()
```

### Single Inheritance 

```python
class Animal:
    def sound(self):
        print("Animal make a sound")

class Dog(Animal):
    def bark(self):
        print("Dpg barks")

d = Dog()

d.sound()
d.bark()
```

### Multiple Inheritance

```python
class Father:
    def money(self):
        print("Father has money")

class Mother:
    def care(self):
        print("Mother takes care")

class Child(Father,Mother):
    def study(self):
        print("Chaild is studying")

c = Child()

c.money()
c.care()
c.study()
        
```

### 3.Multilevel Inheritance

```python
class Grandfather:
    def house(self):
        print("Grandfather owns a house")

class Father(Grandfather):
    def bike(self):
        print("Father own a bike")

class son(Father):
    def laptop(self):
        print("Son own a laptop")

s = son()

s.house()
s.bike()
s.laptop()
    
```

### 4.Hierarchical Inheritance

```python
class Animal:
    def eat(self):
        print("Animal in eating")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meow")

d = Dog()
c = Cat()

d.eat()
d.bark()

c.eat()
c.meow()
```

## 3.Polymorphism

### Definition

* Polymorphism means "one interface, many forms."

* In Object-Oriented Programming (OOP), the **same method name can perform different actions depending** on the object that calls it.

### Simple Meaning
* Poly = Many
* Morphism = Forms

* Polymorphism = Many Forms

### Real-Life Example

#### Think of a person.

* At home → Son/Daughter
* At office → Employee
* With friends → Friend

* The same person behaves differently in different situations.This is polymorphism.

### Example :

```python
class Payment:
    def pay(self):
        print("Payment processing")

class CreditCard(Payment):
    def pay(self):
        print("Paid using Credit card")

class UPI(Payment):
    def pay(self):
        print("Paid using UPI")

class NetBanking(Payment):
    def pay(self):
        print("Paid using Net Banking")

        
payments = [CreditCard(),UPI(),NetBanking()]

for payment in payments:
    payment.pay()
```
