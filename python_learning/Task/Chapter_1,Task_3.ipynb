## Rock, Paper, Scissors Game
### Objective:
- Build a simple Rock-Paper-Scissors game where the player competes with the computer.
- This task will test your understanding of: 
*	Input/output
*	Conditional logic
*	Random module

*	Exception handling

### Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.



```python
import random

h_score = 0
c_score = 0

while True:

    print("\n1 --> Rock")
    print("2 --> Paper")
    print("3 --> Scissors")

    try:
        random_value = random.randint(1,3)
        user_input = int(input("Enter your selecting number"))

        #Check for invalid input
        if user_input< 1 or user_input > 3:
            print("Invalid input!!!,Pleace enter 1,2,or 3")
            continue
        
        print("Computer selecterd",random_value)
    
        if random_value == user_input:
            print("Draw")
        elif random_value == 1 and user_input == 2:
            print("Humam win")
            h_score+=1
        elif random_value == 1 and user_input == 3:
            print("Computer win")
            c_score+=1
        elif random_value == 2 and user_input == 1:
            print("Computer win")
            c_score+=1
        elif random_value == 2 and user_input == 3:
            print("Human win")
            h_score+=1
        elif random_value == 3 and user_input == 1:
            print("Computer win")
            c_score+=1
        elif random_value == 3 and user_input == 2:
            print("Computer win")
            c_score+=1
        

        print("Humen score",h_score)
        print("Computer score",c_score)

        user_input2 = int(input("\nYou continue or break contiune press 1 break press 2"))
        
        if user_input2 == 2:
            print("\nFinal Score")
            print("Human:", h_score)
            print("Computer:", c_score)
            print("Thanks for playing!")
            break
  
    
    except ValueError:
        print("Invalid charecter not recognized.Place enter valid number.!!!")
        
```
