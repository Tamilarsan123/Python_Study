### Data Cleaning Chalenge on sales data

### Import necessary libraries.

```python
import pandas as pd
import numpy as np
```

```python
slaes_data_2017 = pd.read_csv("Sales Transactions-2017.csv")
slaes_data_2018 = pd.read_csv("Sales Transactions-2018.csv")
slaes_data_2019 = pd.read_csv("Sales Transactions-2019.csv")
```

```python
sales_full_data = pd.concat(objs = [slaes_data_2017,slaes_data_2018,slaes_data_2019])
sales_full_data
```

## Perform initial analysis 

```python
sales_full_data.duplicated().sum()
```

```python
sales_full_data.isna().sum()
```

```python
sales_full_data.dtypes
```

```python
sales_full_data.describe(include= 'all')
```

### Step 1: Remove duplicate:

```python
sales_full_data.drop_duplicates(inplace= True)
sales_full_data.head()
```

### Step 2: Fill emptys to null

```python
sales_full_data = sales_full_data.replace(r'^\s*$',np.nan, regex=True)
sales_full_data.head()
```

### Step 3:Convert the Voucher column to the int data type

```python
sales_full_data["Voucher"]=sales_full_data["Voucher"].str.replace("Sal:","",regex=False)#remove the charecter
sales_full_data["Voucher"].head()
```

```python
sales_full_data = sales_full_data.drop(sales_full_data[sales_full_data["Voucher"] =="Total  "].index)# Total charecter can remove in voucher
```

```python
sales_full_data["Voucher"].isna().sum()
```

```python
sales_full_data = sales_full_data.dropna(subset = "Voucher")#Remove the null entery in voucher
sales_full_data
```

```python
sales_full_data["Voucher"] = sales_full_data["Voucher"].astype(int)#Data type change in int
```

```python
sales_full_data.dtypes
```

### Step 4: Convert the Date column to the datetime data type

```python
sales_full_data["Date"]
```

```python
sales_full_data["Date"] = pd.to_datetime(sales_full_data["Date"],format ="%d/%m/%Y")# Data type can change date time and format also set
```

```python
sales_full_data.dtypes
```

### Step 5: Convert all the values in the Party column to uppercase

```python
sales_full_data["Party"] = sales_full_data["Party"].str.upper()
```

```python
sales_full_data.head()
```

### Step 6:Convert all the values in the Product column to uppercase

```python
sales_full_data["Product"] = sales_full_data["Product"].str.upper()
```

```python
sales_full_data.tail()
```

### Step 7:Convert the Qty column to the int data type.

```python
sales_full_data["Qty"].isna().sum()#Check null value
```

```python
sales_full_data["Qty"] = sales_full_data["Qty"].str.replace(",","",regex = True)# This vale 1,00 to chane 1.00

#Use regex=True when you want to match patterns (digits, whitespace, letters, start/end of string, etc.).
#Use regex=False when you want to replace exact text without any special regex behavior.
```

```python
sales_full_data["Qty"] = sales_full_data["Qty"].astype(float).astype(int)#First, change the data type to float, and then convert it to int
```

```python
sales_full_data.dtypes
```

### Step 8:Convert the Rate column to the float data type.

```python
sales_full_data["Rate"]
```

```python
sales_full_data["Rate"] = sales_full_data["Rate"].str.replace(",","") # Replace the str value 1,33.00 to int 133.00
```

```python
sales_full_data["Rate"] = sales_full_data["Rate"].astype(float)# Convert datatype object to float
```

```python
sales_full_data.dtypes
```

### Step 9: Drop the many null value column

**Best practice:** remove a column when it has very high missing values and little business value.

**Common rule:** if a column has > 60–80% null values, many data analysts consider dropping it.

#### 9.1:Drop column 1:Disc

```python
sales_full_data["Disc"].isna().sum()
```

```python
sales_full_data.drop(labels = "Disc",axis = 1,inplace= True)#Drop the colunm Dict
```

```python
sales_full_data.head()
```

#### 9.2:Drop column Voucher Amoune

```python
sales_full_data["Voucher Amount"].isna().sum()
```

```python
sales_full_data.drop(labels ="Voucher Amount",axis=1,inplace=True )#Drop the column Voucher Amount
```

```python
sales_full_data.head()
```

```python
sales_full_data.drop(labels ="Gross",axis=1,inplace=True )
```

```python
sales_full_data
```
