### 1.while loop

```python
password = "tamil"

while True:
    user_input = input("Enter a password:")

    if user_input == password:
        print("Password is currect!!!")
        break
    else:
        print("Wrong Password!!! Enter valid password")
```

```python
balance = 1000

while True:
    print("1.withdraw")
    print("2.Deposit")
    print("3.Balance")
    print("4.Exit")
    
    user_input = int(input("Enter your option:"))

    if user_input == 1:
        amount= int(input("Enter your amount:"))
        balance-=amount
        print("Balance:",balance)

    elif user_input == 2:
        amount= int(input("Enter your amount:"))
        balance+=amount
        print("Balance:",balance)

    elif user_input == 3:
        print("Balance:",balance)

    elif user_input == 4:
        break
```

### 7.user defined functions with for loop and if condition

```python
def prime_number(number):
    count = 0
    for i in range(1,number+1):
        if number%i==0:
            count+=1
    if count==2:
        print("PRIME")
    else:
        print("NOT PRIME")
      
prime_number(7)
```

```python
def count_vowel(word):
    count_vowel     = 0
    count_consonent = 0
    vowel = "aeiou"
    
    for i in word.lower():
        if i in vowel:
            count_vowel+=1
        else:
            count_consonent+=1
    
    print("Count Vowel:",count_vowel)
    print("Count COnsonent:",count_consonent)


count_vowel("tamil")
        
```
