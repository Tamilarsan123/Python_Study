#### 2.set{}

```python
set_1 = {"apple",12,33.2,"orenge"}
```

```python
set_1.add("mango")#add value in set at alphabet
set_1
```

```python
set_1.clear()#retun a empty set
set_1
```

```python
set_2 = set_1.copy()
set_2
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

result = set_3.difference(set_4)#common element presant in secount set that elament remover from first set

print(result)#create a new set
```

```python
set_3.difference_update(set_4)# The difference_update() method removes all element from the original set that are also present in another set.
set_3
```

```python
set_1 = {"apple",12,33.2,"orenge"}
set_1.discard("apple")#Remove element from a set
set_1
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

result = set_3.intersection(set_4)#Print a common elemet from both

print(result)#create a new set
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

set_3.intersection_update(set_4)#common elemet from both set its affect in original set

set_3
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

set_3.isdisjoint(set_4)#The isdisjoint() method checks whether two sets have no common elements.
```

```python
set_3 = {4,5}
set_4 = {4,5,6,7}

#Checks if a set is inside another set
set_3.issubset(set_4)#The issubset() method checks whether all elements of one set are present in another set.
```

```python
set_3 = {1,2,3,4,5}
set_4 = {1,2,3}

#Checks if a set contains another set
set_3.issuperset(set_4)
```

```python
set_1 = {"Apple","Banana", "Mango"}

item = set_1.pop()
#The pop() method removes and returns a random element from a set.
#Since sets are unordered, you cannot predict which element will be removed.

print("Removed:", item)
print("Remaining Set:", set_1)
```

```python
set_1 = {"apple",12,33.2,"orenge"}
set_1.remove("apple")#Remove an element from a set
set_1
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

#The symmetric_difference() method returns a new set containing elements that are in either set, but not in both.

set_3.symmetric_difference(set_4)

#Keeps unique elements from both sets.
#Removes the common elements.
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

#The symmetric_difference_update() method updates the original set with the symmetric difference of itself and another set.

set_3.symmetric_difference_update(set_4)

set_3#Original set will be modified
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

#The union() method returns a new set containing all unique elements from two or more sets.

result = set_3.union(set_4)
print(result)
```

```python
set_3 = {1,2,3,4,5}
set_4 = {4,5,6,7}

set_3.update(set_4)#The update() method adds all unique elements from another iterable (set, list, tuple, string, etc.) to the original set.

set_3
```

```python
students_marks = {"Name":["Siva","Mani","Thaara","Tharun","Viji"],
                 "Tamil":[77,88,97,87,85],
                 "English":[89,76,90,78,76],
                 "Maths":[78,66,97,68,77],
                     }

result = students_marks.popitem()#Remove and return a (key, value) pair as a 2-tuple.
#Pairs are returned in LIFO (last-in, first-out) order.
print("remove:",result)
students_marks
```

```python
students_marks = {"Name":["Siva","Mani","Thaara","Tharun","Viji"],
                 "Tamil":[77,88,97,87,85],
                 "English":[89,76,90,78,76],
                 "Maths":[78,66,97,68,77],
                     }
students_marks.setdefault("Science",[55,78,99,66,88])

#The setdefault() method returns the value of a specified key.
#If the key exists, it returns its value.
#If the key does not exist, it inserts the key with a default value and returns that value

students_marks
```
