### Positional Arguments

```python
def get_details(name,place,expected_price):
    print("Clint Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details(expected_price = 97.5,place = "Chennai",name = "Raj")
```

### Default parameters and arguments

```python
def get_details_chennai(name,place = "Chennai",expected_price):
    print("Clint Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
def get_details_chennai(name,expected_price,place = "Chennai"):
    print("Clint Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details_chennai(expected_price = 97.5,name = "Raj")
```

```python
get_details(expected_price = 97.5,place = "Bangalore",name = "Raj")
```

### Variable-length argument(*)

```python
def get_details_chennai(name,place,*expected_price):
    print("Clint Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details_chennai("Tamil","Chennai",95.5,97.8,99,5) #last parameter gives multible arguments
```

### Keyword Argument(**)

```python
def get_details_chennai(name,place,**expected_price):
    print("Clint Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details_chennai("Tamil","Chennai",ask_1 = 95.5,ask_2 = 93.8,ask_4 = 90.5,ask_3 = 87.5)
```

## 5.4 Challenge-2 Find the Factors of a number

If the user enters 20,output be "The factors of 20 is 1,2,4,5,10,20"

```python
%%time
user_input = int(input("Enter a number"))

fact =[]

for i in range(1,user_input+1):
    if user_input%i==0:
        fact.append(i)
    

print("The Factors of {} is {}".format(user_input,fact))
```
